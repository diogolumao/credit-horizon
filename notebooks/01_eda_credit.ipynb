## 1. Ingestão e Inspeção Inicial
Vamos carregar os dados brutos, preservados na nossa camada `raw`, para entender a volumetria e a completude do dataset.

```python
# Carregamento dos dados
caminho_dados = '../data/raw/german_credit_data.csv'
df = pd.read_csv(caminho_dados)

# Visualizando as primeiras linhas
display(df.head())

# Verificando a estrutura e valores nulos
print("\n--- Informações do Dataset ---")
df.info()
```

## 2. Análise Univariada: O Perfil Demográfico
A primeira pergunta de negócio é: **Quem é o nosso cliente padrão?**
Vamos utilizar nosso módulo visual customizado para analisar a distribuição etária da carteira.

```python
# Utilizando a função da lib local utils/plots.py
fig_idade = plots.plot_histograma_idade(df)
fig_idade.show()
```

**Insight Preliminar:** Notamos uma concentração de clientes em faixas etárias mais jovens. Precisaremos avaliar se a falta de histórico de crédito desta faixa impacta diretamente nas taxas de default (inadimplência).

## 3. Análise Bivariada: Valor, Prazo e Finalidade
Como o tempo e o volume financeiro afetam o risco?

```python
# Relação Valor x Prazo
fig_dispersao = plots.plot_dispersao_amount(df)
fig_dispersao.show()
```

É possível observar uma correlação natural: maiores valores solicitados exigem prazos mais longos. Mas e a finalidade do crédito?

```python
# Boxplot de Finalidade do Empréstimo
fig_purpose = plots.plot_boxplot_purpose(df)
fig_purpose.show()
```

## 4. Conclusões e Direcionamento para Modelagem

A partir desta exploração visual, validamos os principais pontos levantados junto à área de negócios:

1. **O Fator Tempo (Duração):** Prazos mais longos possuem maior variação de risco, especialmente em tickets mais altos.
2. **A Finalidade do Recurso:** Categorias como 'Carro Novo' e 'Eletrodomésticos' possuem perfis de distribuição financeira diferentes de 'Negócios', o que será uma *feature* crucial para o algoritmo.
3. **Próximos Passos:** O dataset apresenta boa integridade e as features selecionadas demonstram variância suficiente para o treinamento de um modelo de **Random Forest**, capaz de capturar essas relações não-lineares.
